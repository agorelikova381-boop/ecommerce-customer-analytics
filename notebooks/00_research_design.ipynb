# 00. Дизайн исследования: Анализ поведения клиентов e-commerce

## 1. Бизнес-задача
**Контекст:** Маркетплейс (на примере Olist) хочет увеличить прибыль. 
**Центральный бизнес-вопрос:** Как увеличить повторные покупки (Retention) и максимизировать выручку за счет понимания поведения разных групп клиентов?

## 2. Декомпозиция на Research Questions (RQ)
*   **RQ1 (Descriptive):** Как выглядят текущие показатели бизнеса? (Выручка, средний чек, топ категорий).
*   **RQ2 (Diagnostic):** Как долго клиенты остаются с бизнесом и когда происходит отток? (Когортный анализ).
*   **RQ3 (Segmentation):** Какие группы клиентов приносят основную прибыль, а какие уходят? (RFM-анализ).

## 3. Дизайн исследования
Поскольку данные уже собраны в виде транзакций, мы проводим **кабинетное исследование (Desk Research)**. Это **количественное исследование (QUANT)**, основанное на анализе вторичных данных.

*   **Метод сбора данных:** Не применимо (используется готовый датасет).
*   **Метод анализа:** 
    *   SQL (DuckDB) для агрегации данных и расчета метрик.
    *   Когортный анализ для оценки удержания (Retention).
    *   RFM-сегментация для классификации клиентской базы.

## 4. Целевая аудитория и выборка
*   **Генеральная совокупность:** Все клиенты маркетплейса.
*   **Целевая аудитория:** Клиенты, совершившие покупку.
*   **Критерии очистки выборки:** Из анализа исключаются заказы со статусом `canceled` и `unavailable` (чтобы не искажать выручку). Анализ строится только на доставленных заказах (`order_status = 'delivered'`).

## 5. Ограничения исследования (Limitations)
*   **Отсутствие маркетинговых данных:** Мы видим факт покупки, но не знаем, какая рекламная кампания предшествовала этому (нет данных о бюджетах и кликах).
*   **География:** Данные бразильского рынка, что может влиять на специфику сезонности и категорий.
*   **Историчность:** Данные 2016–2018 гг.
